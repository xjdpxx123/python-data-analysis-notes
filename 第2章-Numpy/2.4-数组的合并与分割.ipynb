# 2-4 NumPy 数组的合并与分割

合并是把多个数组连接在一起，分割是把一个数组拆成多个子数组。

两种操作都需要关注 **`axis` 指定的轴**。对于二维数组：

| `axis` | 对应维度 | 合并时的变化 |
|---|---|---|
| `0` | 行所在的维度 | 增加行数，列数不变 |
| `1` | 列所在的维度 | 增加列数，行数不变 |

````python
import numpy as np
````

## 1. 数组的合并

常用合并函数如下，本节以二维数组为例：

| 方法 | 作用 | 形状要求 |
|---|---|---|
| `np.concatenate(..., axis=0)` | 按行连接 | 列数相同 |
| `np.concatenate(..., axis=1)` | 按列连接 | 行数相同 |
| `np.vstack()` | 垂直堆叠 | 列数相同 |
| `np.hstack()` | 水平堆叠 | 行数相同 |

准备三个数组，形状分别为 `(2, 3)`、`(1, 3)` 和 `(2, 5)`：

````python
a = np.arange(6).reshape(2, 3)
b = np.arange(3).reshape(1, 3)
c = np.arange(10).reshape(2, 5)
a
````

````python
b
````

````python
c
````

### 1.1 使用 `np.concatenate()` 按指定轴连接

基本格式：

```python
np.concatenate([数组1, 数组2, ...], axis=0)
```

`axis=0` 表示沿第 0 轴连接。二维情况下，可以理解为把 `b` 放到 `a` 下方。

````python
joined_rows = np.concatenate([a, b], axis=0)
joined_rows
````

`a` 和 `b` 的列数都是 3，因此可以连接，结果形状为 `(3, 3)`。

````python
joined_rows.shape
````

`axis=1` 表示沿第 1 轴连接。二维情况下，可以理解为把 `c` 放到 `a` 右侧。

````python
joined_columns = np.concatenate([a, c], axis=1)
joined_columns
````

`a` 和 `c` 的行数都是 2，结果形状为 `(2, 8)`。

````python
joined_columns.shape
````

**形状要求：** 参与 `concatenate()` 的数组维度数应相同；除连接轴之外，其他轴的大小必须一致。

例如，`a` 和 `c` 不能按 `axis=0` 连接，因为列数分别为 3 和 5。

### 1.2 使用 `np.vstack()` 垂直堆叠

`v` 来自 vertical，表示垂直方向。对于本节的二维数组，相当于 `np.concatenate(..., axis=0)`。

````python
np.vstack([a, b])
````

### 1.3 使用 `np.hstack()` 水平堆叠

`h` 来自 horizontal，表示水平方向。对于本节的二维数组，相当于 `np.concatenate(..., axis=1)`。

````python
np.hstack([a, c])
````

> 注意：上述方向说明针对二维数组。一维数组只有第 0 轴，`np.hstack()` 会直接将一维数组首尾连接。

## 2. 数组的分割

使用 `np.split()` 将数组分成多个子数组，返回结果是一个**包含数组的列表**。

基本格式：

```python
np.split(ary, indices_or_sections, axis=0)
```

| 参数 | 含义 |
|---|---|
| `ary` | 要分割的数组 |
| `indices_or_sections` | 整数表示等分份数；索引列表表示切分位置 |
| `axis` | 沿哪个轴分割，默认为 `0` |

### 2.1 按指定索引分割

创建一个包含 0～8 的一维数组，用 `[2, 5]` 指定切分位置。

````python
ary = np.arange(9)
ary
````

````python
parts = np.split(ary, [2, 5])
parts
````

`[2, 5]` 表示在索引 2 和索引 5 **之前**切开，得到三段：

- `ary[:2]`：`[0, 1]`
- `ary[2:5]`：`[2, 3, 4]`
- `ary[5:]`：`[5, 6, 7, 8]`

切分位置是索引，不是“每一段的长度”。结果中的每一段可以像普通列表元素一样取出。

````python
parts[1]
````

### 2.2 按整数等分

`indices_or_sections=3` 表示将该轴等分为 3 份。9 个元素可以分为每份 3 个元素。

````python
np.split(ary, 3)
````

如果份数不能整除指定轴的长度，`np.split()` 会引发 `ValueError`。

例如，`np.split(ary, 2)` 无法将 9 个元素等分为 2 份。若允许各段长度不完全相同，可以使用 `np.array_split()`：

````python
np.array_split(ary, 2)
````

### 2.3 按轴分割二维数组

创建一个 4 行、4 列的数组：

````python
X = np.arange(16).reshape(4, 4)
X
````

`axis=0`：将 4 行分为两组，得到两个形状为 `(2, 4)` 的数组，即上半部分和下半部分。

````python
np.split(X, 2, axis=0)
````

`axis=1`：将 4 列分为两组，得到两个形状为 `(4, 2)` 的数组，即左半部分和右半部分。

````python
np.split(X, 2, axis=1)
````

## 3. 要点总结

- `np.concatenate()` 沿已有的轴连接数组，不会凭空增加维度。
- 对二维数组，`axis=0` 合并增加行数，`axis=1` 合并增加列数。
- `np.vstack()` 和 `np.hstack()` 分别适合二维数组的垂直堆叠和水平堆叠。
- `np.split()` 接收整数时进行等分，指定轴的长度必须能被份数整除。
- `np.split()` 接收索引列表时按指定位置切开，返回由子数组组成的列表。
- 对二维数组，`axis=0` 分割行，`axis=1` 分割列。
