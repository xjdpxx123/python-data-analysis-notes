# 2-5 NumPy 聚合运算

**聚合运算**是把一组数据汇总为一个或少量结果，例如求和、最小值、最大值和平均值。

本节包括：

1. 一维数组的聚合
2. 多维数组按轴聚合
3. 均值与分位数

````python
import numpy as np
````

## 1. 一维数组的聚合

对一维数组直接聚合，通常会得到一个数。

| 函数写法 | 数组方法写法 | 含义 |
|---|---|---|
| `np.sum(a)` | `a.sum()` | 求和 |
| `np.min(a)` | `a.min()` | 求最小值 |
| `np.max(a)` | `a.max()` | 求最大值 |
| `np.mean(a)` | `a.mean()` | 求平均值 |

先创建一维数组：

````python
a = np.arange(9)
a
````

### 1.1 使用 `sum()` 求和

````python
np.sum(a)
````

````python
a.sum()
````

两种写法结果相同，都是 $0 + 1 + \cdots + 8 = 36$。

### 1.2 使用 `min()` 和 `max()` 求最值

````python
np.min(a), np.max(a)
````

````python
a.min(), a.max()
````

结果依次是最小值 `0` 和最大值 `8`。这里返回的是**元素的值**，不是元素所在的索引。

## 2. 多维数组按轴聚合

对二维数组，需要区分“汇总全部元素”和“分别汇总每一行或每一列”。

| `axis` | 聚合方式 | 二维数组的结果 |
|---|---|---|
| 不指定或 `None` | 聚合全部元素 | 一个值 |
| `0` | 沿行所在的维度聚合 | 每一列得到一个结果 |
| `1` | 沿列所在的维度聚合 | 每一行得到一个结果 |

可以记作：**`axis` 指定被汇总掉的维度**。对于形状 `(行数, 列数)`，`axis=0` 汇总各行，保留列；`axis=1` 汇总各列，保留行。

````python
X = np.arange(16).reshape(4, 4)
X
````

### 2.1 不指定 `axis`：汇总所有元素

````python
np.sum(X)
````

结果为 `120`，是数组中全部 16 个元素的和。

### 2.2 `axis=0`：每一列得到一个结果

第 1 列求和为 $0 + 4 + 8 + 12 = 24$，其余列同理。

````python
np.sum(X, axis=0)
````

结果为 `[24, 28, 32, 36]`，依次对应 4 列的和。

### 2.3 `axis=1`：每一行得到一个结果

第 1 行求和为 $0 + 1 + 2 + 3 = 6$，其余行同理。

````python
np.sum(X, axis=1)
````

结果为 `[6, 22, 38, 54]`，依次对应 4 行的和。

### 2.4 沿指定轴求最小值

`np.min(X, axis=0)` 得到每一列的最小值。

````python
np.min(X, axis=0)
````

`np.min(X, axis=1)` 得到每一行的最小值。`max()` 的 `axis` 含义与此相同。

````python
np.min(X, axis=1)
````

## 3. 其他聚合操作

### 3.1 使用 `mean()` 求均值

**均值**就是算术平均数，等于元素之和除以元素个数。

沿 `axis=0` 求均值，得到每一列的均值：

````python
np.mean(X, axis=0)
````

结果为 `[6., 7., 8., 9.]`。例如，第 1 列的均值为 $(0 + 4 + 8 + 12) / 4 = 6$。

沿 `axis=1` 求均值，得到每一行的均值：

````python
np.mean(X, axis=1)
````

结果为 `[1.5, 5.5, 9.5, 13.5]`。不指定 `axis` 时，得到整个数组的均值。

````python
np.mean(X)
````

### 3.2 使用 `percentile()` 求百分位数

**百分位数**描述数据从小到大排列后所处的位置，常用于观察数据分布。

基本格式：

```python
np.percentile(a, q=25)
```

| `q` | 含义 |
|---|---|
| `0` | 最小值 |
| `25` | 第 25 百分位数，即第一四分位数 |
| `50` | 第 50 百分位数，即中位数 |
| `75` | 第 75 百分位数，即第三四分位数 |
| `100` | 最大值 |

`q` 的取值范围为 `0`～`100`，第 25 百分位数应写 `q=25`，不是 `q=0.25`。

对于有限样本，百分位数可能由相邻数值插值得到，不一定恰好是原数组中的某个元素。

````python
np.percentile(np.array([0, 10, 20, 30, 40]), q=25)
````

上例第 25 百分位数为 `10`。再用固定随机种子生成 100 万个位于 `[0, 1)` 的随机数，观察均匀分布的分位数。

`np.random.default_rng(42)` 创建局部随机数生成器；重新从本单元格执行，可以复现同一组示例数据。

````python
rng = np.random.default_rng(42)
random_values = rng.random(1_000_000)
random_values[:5]
````

````python
np.percentile(random_values, q=25)
````

由于数据在 `[0, 1)` 上均匀分布，第 25 百分位数通常接近 `0.25`，但不会要求恰好等于 `0.25`。

````python
np.percentile(random_values, q=100), np.max(random_values)
````

第 100 百分位数与最大值相同。可以使用循环一次查看几个常用的百分位数：

````python
q_list = [0, 25, 50, 75, 100]
for q in q_list:
    value = np.percentile(random_values, q=q)
    print(f"第 {q:3d} 百分位数：{value:.8f}")
````

`q` 也可以直接传入列表，结果会按列表中的顺序返回：

````python
np.percentile(random_values, q=q_list)
````

## 4. 要点总结

- `sum()`、`min()`、`max()`、`mean()` 分别用于求和、最小值、最大值和均值。
- `np.sum(a)` 和 `a.sum()` 是常见的两种调用方式。
- 不指定 `axis` 时，通常对整个数组的所有元素进行聚合。
- 二维数组中，`axis=0` 得到每列的结果，`axis=1` 得到每行的结果。
- `np.percentile()` 的 `q` 范围是 `0`～`100`，`q=50` 对应中位数。
- 随机数据的统计结果具有波动；固定随机种子可以让示例重现。
