# 3-1 Series基础

Pandas 的两种主要数据结构是 `Series` 和 `DataFrame`。

`Series` 是带有索引的一维数据，可以理解为“一列数据 + 对应的标签”。

> 约定：使用 `pd` 作为 Pandas 的别名，使用 `np` 作为 NumPy 的别名。本节可独立从上到下运行。

```python
import pandas as pd
import numpy as np
```

## 1. 创建 Series

| 数据来源 | 写法 | 默认索引 |
|---|---|---|
| 列表 | `pd.Series(列表)` | 从 0 开始的整数索引 |
| NumPy 数组 | `pd.Series(数组)` | 从 0 开始的整数索引 |
| 字典 | `pd.Series(字典)` | 字典的键 |

### 1.1 由列表创建

左侧是索引，右侧是元素值；底部的 `dtype` 表示元素的数据类型。

```python
s1 = pd.Series([1, 2, 5, 6, 8])
s1
```

### 1.2 由 NumPy 数组创建

```python
s2 = pd.Series(np.array(["a", "b", "c", "d"]))
s2
```

字符串类型的具体显示名称可能随 Pandas 版本不同而变化。

### 1.3 由字典创建

字典的键成为索引，字典的值成为 Series 的数据。

```python
s3 = pd.Series({"a": 1, "b": 2, "c": 3})
s3
```

## 2. 指定索引与查看数据

`index` 的长度必须与数据长度一致。

```python
s4 = pd.Series([1, 2, 3, 4, 5], index=["a", "b", "c", "d", "e"])
s4
```

```python
s4.index, s4.values
```

`index` 查看索引，`values` 查看底层数据。需要明确转换为 NumPy 数组时，也可以使用 `to_numpy()`。

```python
s4.to_numpy()
```

标签访问使用 `loc`，位置访问使用 `iloc`，两种意图不要混淆。

```python
s4.loc["b"], s4.iloc[1]
```

结果都是 `2`：标签 `b` 恰好位于第 2 个位置。

## 3. Series 的运算

### 3.1 与单个数运算

加、减、乘、除会作用于每一个元素，索引保持不变。

```python
pd.DataFrame({"原数据": s4, "加1": s4 + 1, "减1": s4 - 1, "乘2": s4 * 2, "除2": s4 / 2})
```

### 3.2 使用条件筛选

`s4 > 3` 生成布尔 Series；用它索引原数据，可以保留满足条件的元素。

```python
s4 > 3
```

```python
s4[s4 > 3]
```

结果保留标签 `d`、`e` 和数值 `4`、`5`；若只要值，可以继续调用 `to_numpy()`。

### 3.3 两个 Series 运算

Series 之间按**索引标签对齐**，不是简单按当前位置相加。

```python
ser_a = pd.Series([1, 2, 3], index=["a", "b", "c"])
ser_b = pd.Series([4, 5, 6], index=["b", "c", "d"])
ser_a + ser_b
```

标签 `b` 的结果为 `2 + 4 = 6`，标签 `c` 的结果为 `3 + 5 = 8`。`a`、`d` 在另一侧没有对应值，因此结果为缺失值 `NaN`。需要把缺少的值按 0 处理时，可显式指定填充值。

```python
ser_a.add(ser_b, fill_value=0)
```

## 4. 要点总结

- Series 是带索引的一维数据，可由列表、数组、字典创建。
- `index` 查看索引，`to_numpy()` 取得数组。
- 标量运算作用于每个元素，条件筛选保留满足条件的标签和值。
- Series 间运算按标签对齐，未匹配的标签通常产生缺失值。
