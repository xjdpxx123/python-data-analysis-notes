# 3-3 Series与DataFrame的关系

可以把 DataFrame 理解为多个共用行索引的 Series 组合而成的表格。

> 约定：使用 `pd` 作为 Pandas 的别名，使用 `np` 作为 NumPy 的别名。本节可独立从上到下运行。

```python
import pandas as pd
import numpy as np
```

## 1. 对比两种数据结构

| 对比项 | Series | DataFrame |
|---|---|---|
| 维度 | 一维 | 二维 |
| 标签 | 一组索引 | 行索引和列名 |
| 形状示例 | `(3,)` | `(3, 3)` |
| 常见用途 | 一列数据 | 一张表格 |

```python
df = pd.DataFrame({
    "name": ["lemon", "Jack", "Jason"],
    "age": [20, 25, 27],
    "sex": ["male", "female", "male"]
})
df
```

## 2. 从 DataFrame 中取出 Series

一层方括号加单个列名返回 Series；传入列名列表会保留二维 DataFrame。

```python
name = df["name"]
age = df["age"]
name
```

```python
type(df["name"]), type(df[["name"]])
```

```python
df["name"].shape, df[["name"]].shape
```

结果分别是 `(3,)` 与 `(3, 1)`。Series 的 `name` 属性可以保存列名称，但它没有 DataFrame 的列索引。

```python
name.name, name.index
```

## 3. 多个 Series 组合为 DataFrame

用列表构造表格后再转置：此时每个 Series 先成为一行，转置后再成为一列。

```python
pd.DataFrame([name, age]).T
```

也可以直接把 Series 放入字典，键指定列名。这种方式更直观，也更容易保留各列独立的数据类型。

```python
combined = pd.DataFrame({"name": name, "age": age})
combined
```

构造时仍然按索引对齐；下例故意改变年龄的索引，便于观察缺失值。

```python
other_age = pd.Series([20, 25, 27], index=[1, 2, 3])
pd.DataFrame({"name": name, "age": other_age})
```

索引 0 没有年龄，索引 3 没有姓名。相同长度不代表位置自动匹配。

## 4. 逐行读取数据

`iterrows()` 每次返回“行索引 + 行 Series”。使用列名取值比依赖列的排列顺序更清楚。

```python
for index, row in df.iterrows():
    print(index, row["age"], row["sex"])
```

结果依次输出三个索引及对应的年龄、性别。逐行遍历适合讲解或少量特殊处理；常规计算优先直接操作整列。不要通过修改遍历得到的 `row` 来更新原表。

```python
df["age"] + 1
```

若确实需要遍历，还可以使用 `itertuples()`；它按行返回命名元组，通常比 `iterrows()` 更高效。

```python
for row in df.itertuples(index=False):
    print(row.name, row.age)
```

## 5. 要点总结

- `df["列名"]` 返回 Series，`df[["列名"]]` 返回 DataFrame。
- 多个 Series 可以组成 DataFrame，组合时会按索引对齐。
- `iterrows()` 逐行返回索引与 Series；一般计算优先使用整列运算。
