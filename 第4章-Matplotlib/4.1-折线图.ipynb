# 4-1 折线图

折线图把一组数据点按顺序连接起来，适合观察随时间或其他有序变量变化的趋势。

本节学习三类操作：

1. 绘制一条或多条折线
2. 设置颜色、宽度、线型与图例
3. 添加标题、坐标轴说明、显示范围与网格

> 约定：使用 `np` 作为 NumPy 的别名，使用 `plt` 作为 `matplotlib.pyplot` 的别名。
> 下方准备单元格会选择可用的中文字体；可以从项目根目录或本章目录运行。
> `fonts` 文件夹是可选的，没有该文件夹时会查找系统字体。
> 如果没有可用的中文字体，会显示提示并使用默认字体继续绘图，中文可能显示为方框。
> 字体设置只作用于当前 Python 进程，详细原理见第 4-6 节。

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import font_manager

# fonts 文件夹是可选的；没有字体文件时，使用系统已安装的字体。
project_dir = Path('.')
if Path.cwd().name == '第4章-Matplotlib':
    project_dir = Path('..')
font_files = [
    project_dir / 'fonts' / 'NotoSansCJKsc-Regular.otf',
    project_dir / 'fonts' / 'NotoSansSC-Regular.ttf',
]
font_path = next((path for path in font_files if path.is_file()), None)

if font_path is None:
    font_candidates = [
        'Microsoft YaHei', 'SimHei', 'Noto Sans CJK SC', 'Noto Sans SC',
        'Source Han Sans SC', 'PingFang SC', 'Heiti SC', 'Songti SC',
        'Arial Unicode MS',
    ]
    available_fonts = {entry.name for entry in font_manager.fontManager.ttflist}
    font_name = next((name for name in font_candidates if name in available_fonts), None)
    if font_name is None:
        font_name = 'DejaVu Sans'
        print('未找到中文字体，暂用 DejaVu Sans，图中的中文可能显示为方框。')
        print('可安装 Noto Sans CJK SC，或将中文字体文件放入项目的 fonts 文件夹后重新运行。')
    font_path = Path(font_manager.findfont(
        font_manager.FontProperties(family=font_name), fallback_to_default=False
    ))
else:
    font_name = font_manager.FontProperties(fname=str(font_path)).get_name()

font_manager.fontManager.addfont(str(font_path))
plt.rcParams['font.family'] = [font_name]
plt.rcParams['axes.unicode_minus'] = False
```

## 1. 准备数据

下面使用年份和两组销量作为示例数据。每组纵坐标的元素个数都必须与横坐标一致。

```python
x_data = [2011, 2012, 2013, 2014, 2015, 2016, 2017]
y_data = [58000, 60200, 63000, 71000, 84000, 90500, 107000]
y_data2 = [52000, 54200, 51500, 58300, 56800, 59500, 62700]
```

## 2. 使用 `plt.plot()` 绘制折线

基本格式为 `plt.plot(x, y)`。在同一个图中连续调用两次，就可以绘制两条折线。

- `plt.figure(figsize=(7, 4))`：创建一张宽 7 英寸、高 4 英寸的图。
- `plt.show()`：显示图像；放在绘图单元格末尾可以避免显示绘图对象的额外文字。

```python
plt.figure(figsize=(7, 4))
plt.plot(x_data, y_data)
plt.plot(x_data, y_data2)
plt.xlabel('年份')
plt.ylabel('销量')
plt.title('两组销量的变化趋势')
plt.tight_layout()
plt.show()
```

## 3. 设置折线样式

| 参数 | 作用 | 示例 |
|---|---|---|
| `color` | 折线颜色 | `'red'`、`'black'` |
| `linewidth` | 折线宽度，单位为磅 | `3` |
| `linestyle` | 折线样式 | `'--'` |
| `marker` | 数据点的标记，可选 | `'o'` |

### 3.1 指定颜色和宽度

将第一条折线设置为红色，第二条设置为黑色，线宽都设置为 `3`。

```python
plt.figure(figsize=(7, 4))
plt.plot(x_data, y_data, color='red', linewidth=3)
plt.plot(x_data, y_data2, color='black', linewidth=3)
plt.xlabel('年份')
plt.ylabel('销量')
plt.title('设置折线颜色与宽度')
plt.tight_layout()
plt.show()
```

### 3.2 指定线型

| `linestyle` | 含义 |
|---|---|
| `'-'` | 实线，默认值 |
| `'--'` | 虚线 |
| `'-.'` | 点划线 |
| `':'` | 点线 |

不同线型可以帮助区分多组数据，黑白打印时也更容易辨认。

```python
plt.figure(figsize=(7, 4))
plt.plot(x_data, y_data, color='red', linewidth=3, linestyle='--')
plt.plot(x_data, y_data2, color='black', linewidth=3, linestyle=':')
plt.xlabel('年份')
plt.ylabel('销量')
plt.title('使用不同线型')
plt.tight_layout()
plt.show()
```

## 4. 添加图例、标题和坐标轴说明

| 方法或参数 | 作用 |
|---|---|
| `label='产品 A'` | 给当前折线命名 |
| `plt.legend()` | 显示各条折线的图例 |
| `plt.xlabel()` | 设置横坐标轴说明 |
| `plt.ylabel()` | 设置纵坐标轴说明 |
| `plt.title()` | 设置图表标题 |

只有设置 `label` 还不够，需要调用 `plt.legend()` 才会显示图例。

```python
plt.figure(figsize=(7, 4))
plt.plot(x_data, y_data, color='red', linewidth=3,
         linestyle='--', marker='o', label='产品 A')
plt.plot(x_data, y_data2, color='black', linewidth=3,
         linestyle=':', marker='o', label='产品 B')
plt.xlabel('年份')
plt.ylabel('销量')
plt.title('2011—2017 年产品销量')
plt.legend()
plt.tight_layout()
plt.show()
```

## 5. 设置坐标轴范围与网格

`plt.axis([x_min, x_max, y_min, y_max])` 按顺序设置横轴和纵轴范围。

| 网格写法 | 作用 |
|---|---|
| `plt.grid(True)` | 显示横轴和纵轴的网格 |
| `plt.grid(True, axis='x')` | 显示对应横轴刻度的竖向网格线 |
| `plt.grid(True, axis='y')` | 显示对应纵轴刻度的横向网格线 |

`plt.grid()` 不传参数时会切换网格的显示状态；明确传入 `True` 可以确保开启。

下面只显示 2013—2016 年附近的曲线，并添加纵轴网格。
设置显示范围不会删除原始数据，但超出范围的部分会被裁掉。

```python
plt.figure(figsize=(7, 4))
plt.plot(x_data, y_data, color='red', linewidth=3, linestyle='--', label='产品 A')
plt.plot(x_data, y_data2, color='black', linewidth=3, linestyle=':', label='产品 B')
plt.xlabel('年份')
plt.ylabel('销量')
plt.title('2013—2016 年销量对比')
plt.axis([2013, 2016, 20000, 150000])
plt.xticks([2013, 2014, 2015, 2016])
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.legend()
plt.tight_layout()
plt.show()
```

## 6. 要点总结

- `plt.plot(x, y)` 绘制折线，横纵坐标的数据长度应一致。
- 连续调用 `plt.plot()` 可以在一张图中对比多组数据。
- `color`、`linewidth`、`linestyle` 分别设置颜色、宽度和线型。
- `label` 配合 `plt.legend()` 显示图例。
- 图表应标清标题、横轴和纵轴；网格有助于读取数值。
- `plt.axis()` 控制显示范围，`plt.show()` 显示当前图表。
